## Step 4. Run the sampling machine
This is the warehouse from the Module 1 animation and the Sample Factory game. Order times average **12 minutes** with a standard deviation of **9 minutes**, and they are skewed right: most orders are quick, and a few take a very long time.

The cell below draws **1,000 samples** of `n` orders each, computes each sample's mean, and draws a histogram of the 1,000 means.

1. Run it with `n = 4`. Look at the shape.
2. Change `n` to `36` and run it again.

The cell remembers both runs for your receipt.

```python
n = 4   # CHANGE THIS: run once with 4, then once with 36

# --- simulation (you do not need to read below this line) ---
if not isinstance(n, int) or n < 1 or n > 1000:
    print("n should be a whole number between 1 and 1000, like 4 or 36.")
else:
    seed = 2620 + int(LAST_DIGIT) * 1000 + n
    rng = np.random.default_rng(seed)
    orders = 3 + rng.exponential(scale=9, size=(1000, n))   # 1,000 samples of n orders
    means = orders.mean(axis=1)                              # one x-bar per sample
    se_theory = 9 / np.sqrt(n)

    fig, ax = plt.subplots(figsize=(7, 3.5))
    ax.hist(means, bins=40, color="#009E73", edgecolor="black", hatch="..")
    ax.axvline(12, color="black", linestyle="--", label="population mean, mu = 12")
    ax.set_xlabel("Sample mean, x-bar (minutes)")
    ax.set_ylabel("Number of samples")
    ax.set_title(f"1,000 sample means, n = {n}")
    ax.legend()
    plt.show()

    runs[n] = (means.mean(), means.std(ddof=1))
    print(f"Center of the 1,000 sample means: {means.mean():.2f} minutes (theory: 12)")
    print(f"Spread of the 1,000 sample means: {means.std(ddof=1):.2f} minutes (theory: 9 / sqrt({n}) = {se_theory:.2f})")
    if n < 30:
        print("Look at the right tail. With a small n, the pile keeps some of the population's skew.")
    else:
        print("The pile is narrow and close to a bell curve: the Central Limit Theorem at work.")
```
